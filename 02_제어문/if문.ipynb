```python
# 만약 3000원 이상의 돈을 가지고 있으면 택시를 타고 가고,
# 그렇지 않으면 걸어가라.

money = 3000

if money >= 3000:
    print("택시를 타고 가라")
else:
    print("걸어가라")
```

```python
# 만약에 돈이 3000원 이상 있거나 카드가 있다면 택시를 타고 가고,
# 그렇지 않으면 걸어가라

money = 3000
card  = True

if money >= 3000 or card:  # card == True:도 가능
    print("택시를 타고 가라")
else:
    print("걸어가라")
    
```

```python
# 만약에 주머니에 돈이 있으면 택시를 타고 가고, 없으면 걸어가라

주머니 = ["paper", "cellphone", "money"]

if "money" in 주머니:
    print("택시를 타고 가라")
else:
    print("걸어가라")
```

```python
# 철수 = 80
# 90점 이상이면 A, 80점 이상이면 B, 70점 이상이면, 60점 이상이면 D, 그 외 F

score = 80

if score >= 90:
    print("A")
elif score >= 80:
    print("B")
elif score >= 70:
     print("C")
elif score >= 60:
     print("D")
else:
     print("F")
```
