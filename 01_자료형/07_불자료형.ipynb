```python
bool(1 == 1)
print(bool("python")) # 문자형은 모두 True
print(bool("")) # 아무것도 없는 문자형은 False
print(bool(10))
print(bool(0))
print(bool([1, 2, 3]))
print(bool([]))
```

```python
## 논리 연산자

# and
# or

print(True and True)
print(True and False)
print(False and True)
print(False and False)

print(True or True)
print(True or False)
print(False or True)
print(False or False)


## 비교연산자
## >(크다, 초과)
## <(작다, 미만)
## >=(크거나 같다, 이상)
## <=(작거나 같다, 이상)
## !=(다르다)
## ==(같다)
```
