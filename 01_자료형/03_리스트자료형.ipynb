```python
# 리스트는 어떻게 만들고 사용할까?
# 리스트명 = [요소1, 요소2, ... ]

a = []
b = [1, 2, 3]
c = ["Life", "is", "too", "short"]
d = [1, 2, ["life", "is"]]
e = [1, 2, "life", "is"]
f = list()

print(a)
print(b)
print(c)
print(d)
print(e)
print(f)
```

```python
## 리스트 관련 함수

## 리스트에 요소 추가하기 - append
a = [1, 2, 3]
a.append(4)
print(a)
```

```python
a.append(["a", "b"])
print(a)
```

```python
## 리스트 정령 = sort

a = [1, 4, 3, 2]
a.sort() # 오름차순
a
```

```python
a = ["a", "c", "b"]
a.sort(reverse=True) # 내림차순
a
```

```python
## 리스트 뒤집기 - reverse

a = ["a", "c", "b"]
a.reverse()
a
```

```python
## 인덱스 반환 - index

a = [1, 2, 3, 4]
a.index(3)
```

```python
## 리스트에 요소 삽입 - insert

a = [1, 2, 3]
a.insert(0,4)
a
```

```python
## 리스트에 요소 제거 - remove

a = [1, 2, 3, 1, 2, 3]
a.remove(2)
a.remove(2)
a
```

```python
## 리스트 요소 꺼내기 - pop

a = [1, 2, 3]
number = a.pop() # 빈칸일 경우 인덱스 -1이 기본값

print(number)
print(a)
```

```python
## 리스트에 포함된 요소 X의 개수 세기 - count

a = [1, 2, 3, 4, 3] 

a.count(3)
```

```python
## 리스트 확장 - extend

a = [1, 2, 3]
a.extend([4, 5])
a
```
